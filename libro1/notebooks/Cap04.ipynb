### 4.6 Caso aplicado: reglas para detectar pedidos sospechosos

````python
"""Reglas para clasificar el riesgo de un pedido de la tienda en línea."""

# --- Datos del pedido que se va a revisar ---
total_pedido = 1850000
hora_del_pedido = 3          # formato de 24 horas: 3 = 3 de la madrugada
ciudad_entrega = "bogota"
ciudad_facturacion = "cartagena"
cliente_nuevo = True
pedidos_ultima_hora = 4

# --- Reglas del negocio (constantes en MAYÚSCULAS) ---
MONTO_ALTO = 1000000
HORA_INICIO_MADRUGADA = 0
HORA_FIN_MADRUGADA = 5
MAXIMO_PEDIDOS_POR_HORA = 2

# --- Cada señal de riesgo es un valor de sí o no ---
es_monto_alto = total_pedido >= MONTO_ALTO
es_de_madrugada = (HORA_INICIO_MADRUGADA <= hora_del_pedido
                   <= HORA_FIN_MADRUGADA)
ciudades_distintas = ciudad_entrega != ciudad_facturacion
muchos_pedidos = pedidos_ultima_hora > MAXIMO_PEDIDOS_POR_HORA

# --- Contamos las señales: en Python, True vale 1 y False vale 0 ---
total_senales = (es_monto_alto + es_de_madrugada
                 + ciudades_distintas + muchos_pedidos)

# --- Decisión ---
if total_senales >= 3 or (total_senales >= 2 and cliente_nuevo):
    decision = "DETENER el pedido y revisarlo a mano"
elif total_senales == 2:
    decision = "PEDIR al cliente que confirme su identidad"
elif total_senales == 1:
    decision = "APROBAR y vigilar"
else:
    decision = "APROBAR"

print(f"Señales de riesgo: {total_senales} de 4")
print(f"Cliente nuevo: {cliente_nuevo}")
print(f"Decisión: {decision}")
````

*Salida en el libro:*

```text
Señales de riesgo: 4 de 4
Cliente nuevo: True
Decisión: DETENER el pedido y revisarlo a mano
```

### [Ex-L1-C4-01] — Empaque y despacho

````python
# --- Datos del despacho ---
pares_por_despachar = 875
PARES_POR_CAJA = 36
COSTO_ENVIO_CAJA = 12000
COSTO_ENVIO_BOLSA = 5000
TASA_IVA = 0.19

# --- 1. Cajas completas y pares sueltos ---
cajas_completas = pares_por_despachar // PARES_POR_CAJA
pares_sueltos = pares_por_despachar % PARES_POR_CAJA
print(f"Cajas completas: {cajas_completas}")
print(f"Pares sueltos: {pares_sueltos}")

# --- 2. Costo del envío: cada caja más una bolsa para los sueltos ---
costo_envio = cajas_completas * COSTO_ENVIO_CAJA + COSTO_ENVIO_BOLSA
print(f"Costo del envío sin IVA: {costo_envio} pesos")

# --- 3. Costo con IVA: el costo más el 19 % del costo ---
costo_con_iva = costo_envio * (1 + TASA_IVA)
print(f"Costo del envío con IVA: {costo_con_iva:.0f} pesos")
````

*Salida en el libro:*

```text
Cajas completas: 24
Pares sueltos: 11
Costo del envío sin IVA: 293000 pesos
Costo del envío con IVA: 348670 pesos
```

````python
# INCORRECTO: sin paréntesis, Python multiplica primero 5000 * 1.19
# y solo le aplica el IVA a la bolsa, no a las cajas.
costo_mal = cajas_completas * COSTO_ENVIO_CAJA + COSTO_ENVIO_BOLSA * 1.19
print(f"Sin paréntesis (mal): {costo_mal:.0f} pesos")

# CORRECTO: los paréntesis hacen que el IVA se aplique a todo el envío.
costo_bien = (cajas_completas * COSTO_ENVIO_CAJA + COSTO_ENVIO_BOLSA) * 1.19
print(f"Con paréntesis (bien): {costo_bien:.0f} pesos")
````

*Salida en el libro:*

```text
Sin paréntesis (mal): 293950 pesos
Con paréntesis (bien): 348670 pesos
```

### [Ex-L1-C4-03] — Semáforo de calidad de un pedido

````python
# --- Pedido que se va a revisar ---
precio = 45000
cantidad = 3

# --- Límite para cantidades raras en una tienda pequeña ---
CANTIDAD_MAXIMA_NORMAL = 100

# --- Semáforo: primero lo más grave ---
if precio < 0 or cantidad <= 0:
    color = "ROJO"
    motivo = "el pedido tiene un dato imposible (precio o cantidad)"
elif cantidad > CANTIDAD_MAXIMA_NORMAL:
    color = "AMARILLO"
    motivo = "la cantidad es muy alta para una tienda pequeña"
else:
    color = "VERDE"
    motivo = "los datos son razonables"

print(f"Precio {precio}, cantidad {cantidad}: {color}, {motivo}.")
````

*Salida en el libro:*

```text
Precio 45000, cantidad 3: VERDE, los datos son razonables.
```

### [Ex-L1-C4-04] — Clasificar clientes según lo que compran

````python
# --- Datos del cliente ---
gasto_anual = 2500000
numero_compras = 12

# --- Reglas del negocio ---
GASTO_VIP = 2000000
COMPRAS_FRECUENTE = 10

# --- Clasificación: primero el error, después de la categoría más
# exigente a la menos exigente ---
if gasto_anual < 0:
    categoria = "Error: gasto imposible"
elif gasto_anual >= GASTO_VIP and numero_compras >= COMPRAS_FRECUENTE:
    categoria = "VIP"
elif numero_compras >= COMPRAS_FRECUENTE:
    categoria = "Frecuente"
elif numero_compras >= 1:
    categoria = "Ocasional"
else:
    categoria = "Inactivo"

print(f"Gasto {gasto_anual}, compras {numero_compras}: {categoria}")
````

*Salida en el libro:*

```text
Gasto 2500000, compras 12: VIP
```
