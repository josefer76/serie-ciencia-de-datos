### 4.1 Operadores aritméticos, de asignación y precedencia

```python
# Operadores aritméticos fundamentales.
suma = 10 + 3            # 13
resta = 10 - 3           # 7
producto = 10 * 3        # 30
division = 10 / 3        # 3.333... -> SIEMPRE devuelve float
division_entera = 10 // 3  # 3  -> descarta la parte decimal
modulo = 10 % 3          # 1   -> resto de la división
potencia = 10 ** 3       # 1000 -> exponenciación
```

```python
# Ejemplo con sentido de negocio: repartir transacciones en lotes.
total_transacciones = 1502
tamano_lote = 100
 
lotes_completos = total_transacciones // tamano_lote  # 15 lotes llenos
transacciones_sobrantes = total_transacciones % tamano_lote  # 2 sobrantes
 
print(f"Lotes completos: {lotes_completos}, "
      f"sobrantes: {transacciones_sobrantes}")
```

### 4.2 Operaciones con strings: el preludio del procesamiento de texto

```python
texto_crudo = "  Transacción APROBADA  "
 
# .strip() elimina espacios sobrantes al inicio y final.
texto_limpio = texto_crudo.strip()        # "Transacción APROBADA"
 
# .lower() y .upper() cambian el caso.
print(texto_limpio.lower())               # "transacción aprobada"
 
# .replace() sustituye subcadenas; .split() divide en lista.
print(texto_limpio.replace("APROBADA", "RECHAZADA"))
print(texto_limpio.split())               # ['Transacción', 'APROBADA']
```

### 4.3 Operadores de comparación y la comparación de flotantes

```python
# La trampa clásica de la aritmética de punto flotante.
print(0.1 + 0.2 == 0.3)  # False (!) -> 0.1 + 0.2 da 0.30000000000000004
 
import math
# Forma CORRECTA de comparar flotantes: con una tolerancia.
suma = 0.1 + 0.2
print(math.isclose(suma, 0.3))  # True
```

### 4.4 Operadores lógicos: and, or, not

```python
monto_transaccion = 1500.0
es_pais_riesgo = True
cliente_verificado = False
 
# Combinamos varias condiciones para una decisión más matizada.
transaccion_sospechosa = (monto_transaccion > 1000.0) and es_pais_riesgo
print(transaccion_sospechosa)  # True
 
# 'not' invierte: marca al cliente NO verificado.
requiere_revision_manual = transaccion_sospechosa and not cliente_verificado
print(requiere_revision_manual)  # True
```

### 4.5 Condicionales: if, elif, else y la indentación obligatoria

```python
# Clasificación del nivel de riesgo de una transacción según su monto.
monto_transaccion = 7500.0
 
if monto_transaccion >= 10000.0:
    nivel_riesgo = "crítico"
elif monto_transaccion >= 5000.0:
    nivel_riesgo = "alto"
elif monto_transaccion >= 1000.0:
    nivel_riesgo = "medio"
else:
    nivel_riesgo = "bajo"
 
print(f"Nivel de riesgo: {nivel_riesgo}")  # Nivel de riesgo: alto
```

### 4.6 Caso aplicado: motor de reglas para detección de transacciones sospechosas

```python
"""
Motor de reglas para clasificar el riesgo de fraude de una transacción.
Combina operadores aritméticos, de comparación y lógicos con condicionales.
"""
 
# --- Datos de entrada de la transacción a evaluar ---
monto_transaccion = 8500.0
hora_del_dia = 3                 # formato 24h; 3 = 3 de la madrugada
pais_origen = "interno"
pais_destino = "externo"
cliente_verificado = False
transacciones_ultima_hora = 6
 
# --- Constantes de política (en MAYUSCULAS según PEP 8) ---
UMBRAL_MONTO_ALTO = 5000.0
HORA_INICIO_MADRUGADA = 0
HORA_FIN_MADRUGADA = 5
MAX_TRANSACCIONES_HORA = 4
 
# --- Evaluación de señales de riesgo individuales (cada una es booleana) ---
es_monto_elevado = monto_transaccion >= UMBRAL_MONTO_ALTO
es_horario_atipico = (HORA_INICIO_MADRUGADA <= hora_del_dia
                      <= HORA_FIN_MADRUGADA)
es_transferencia_externa = ((pais_origen == "interno")
                            and (pais_destino == "externo"))
es_frecuencia_anomala = transacciones_ultima_hora > MAX_TRANSACCIONES_HORA
 
# --- Conteo de señales activas (en Python True vale 1 y False vale 0) ---
total_senales = (
    es_monto_elevado
    + es_horario_atipico
    + es_transferencia_externa
    + es_frecuencia_anomala
)
 
# --- Lógica de decisión combinada ---
factor_agravante = not cliente_verificado
 
if total_senales >= 3 or (total_senales >= 2 and factor_agravante):
    decision = "BLOQUEAR y escalar a revisión manual"
elif total_senales == 2:
    decision = "RETENER y solicitar verificación adicional"
elif total_senales == 1:
    decision = "APROBAR con monitoreo reforzado"
else:
    decision = "APROBAR"
 
print(f"Señales de riesgo detectadas: {total_senales} de 4")
print(f"Cliente verificado: {cliente_verificado}")
print(f"Decisión del motor: {decision}")
```

### [Ex-L1-C4-01] — Calculadora y clasificadora de IMC

```python
"""Calculadora de IMC con clasificación según los rangos estándar de la OMS."""
 
# --- Constantes de umbrales (MAYUSCULAS, PEP 8) ---
UMBRAL_BAJO_PESO = 18.5
UMBRAL_NORMAL = 25.0
UMBRAL_SOBREPESO = 30.0
 
# --- Entrada de datos y conversión de tipo ---
peso_kg = float(input("Introduce tu peso en kilogramos: "))
estatura_m = float(input("Introduce tu estatura en metros: "))
 
# --- Cálculo del IMC: peso / estatura^2 ---
# El operador ** eleva al cuadrado la estatura.
indice_masa_corporal = peso_kg / (estatura_m ** 2)
 
# Mostramos el IMC redondeado a un decimal con f-string.
print(f"Tu IMC es: {indice_masa_corporal:.1f}")
 
# --- Clasificación multi-rama ---
# El orden importa: evaluamos de menor a mayor, capturando cada rango una vez.
if indice_masa_corporal < UMBRAL_BAJO_PESO:
    categoria = "bajo peso"
    recomendacion = "Considera consultar a un profesional sobre tu nutrición."
elif indice_masa_corporal < UMBRAL_NORMAL:
    # Si llegamos aquí, ya NO es < 18.5, así que está en [18.5, 25).
    categoria = "peso normal"
    recomendacion = "Tu peso está en un rango saludable. ¡Sigue así!"
elif indice_masa_corporal < UMBRAL_SOBREPESO:
    # Aquí sabemos que está en [25, 30).
    categoria = "sobrepeso"
    recomendacion = "Pequeños cambios en hábitos pueden ser beneficiosos."
else:
    # Todo lo restante: IMC >= 30.
    categoria = "obesidad"
    recomendacion = ("Te recomendamos orientación de un profesional "
                     "de la salud.")
 
# --- Salida final ---
print(f"Categoría: {categoria}.")
print(recomendacion)
```

### [Ex-L1-C4-02] — Validador de robustez de contraseñas

```python
"""Validador de robustez de contraseñas usando solo métodos de string y
lógica booleana."""
 
# --- Constantes de política ---
LONGITUD_MINIMA = 8
CARACTERES_ESPECIALES = "!@#$%&*"
 
# --- Entrada ---
contrasena = input("Introduce una contraseña para evaluar: ")
 
# --- Criterio 1: longitud mínima ---
cumple_longitud = len(contrasena) >= LONGITUD_MINIMA
 
# --- Criterio 2: contiene al menos una mayúscula ---
# Truco sin bucles: si la versión en minúsculas DIFIERE de la original,
# es porque había al menos una mayúscula que .lower() transformó.
tiene_mayuscula = contrasena.lower() != contrasena
 
# --- Criterio 3: contiene al menos una minúscula (complementario) ---
tiene_minuscula = contrasena.upper() != contrasena
 
# --- Criterio 4: contiene al menos un dígito ---
# Sumamos los conteos de cada dígito 0-9. .count() recorre internamente.
cantidad_digitos = (
    contrasena.count("0") + contrasena.count("1") + contrasena.count("2")
    + contrasena.count("3") + contrasena.count("4") + contrasena.count("5")
    + contrasena.count("6") + contrasena.count("7") + contrasena.count("8")
    + contrasena.count("9")
)
tiene_digito = cantidad_digitos > 0
 
# --- Criterio 5: contiene al menos un carácter especial ---
cantidad_especiales = (
    contrasena.count("!") + contrasena.count("@") + contrasena.count("#")
    + contrasena.count("$") + contrasena.count("%") + contrasena.count("&")
    + contrasena.count("*")
)
tiene_especial = cantidad_especiales > 0
 
# --- Conteo de criterios de complejidad cumplidos ---
# En Python True vale 1 y False vale 0, así que se suman.
criterios_complejidad = (
    tiene_mayuscula + tiene_minuscula + tiene_digito + tiene_especial
)
 
# --- Lógica de decisión con cortocircuito (reto adicional) ---
# Si NO cumple la longitud mínima, es débil de inmediato.
if not cumple_longitud:
    robustez = "débil (no alcanza la longitud mínima)"
elif criterios_complejidad >= 4:
    robustez = "fuerte"
elif criterios_complejidad >= 2:
    robustez = "media"
else:
    robustez = "débil"
 
# --- Informe ---
print(f"Longitud suficiente: {cumple_longitud}")
print(f"Criterios de complejidad cumplidos: {criterios_complejidad} de 4")
print(f"Robustez de la contraseña: {robustez}")
```
