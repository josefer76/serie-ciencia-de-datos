### 6.1 Listas: datos en orden, que se pueden cambiar

#### Leer un elemento por su posición

````python
productos = ["camisa", "jean", "chaqueta", "medias"]

print(productos[0])    # el primero
print(productos[2])    # el tercero
print(productos[-1])   # el último (los negativos cuentan desde el final)
````

*Salida en el libro:*

```text
camisa
chaqueta
medias
```

#### Tomar un pedazo de la lista

````python
productos = ["camisa", "jean", "chaqueta", "medias"]

print(productos[0:2])   # posiciones 0 y 1
print(productos[:3])    # desde el principio hasta antes de la posición 3
print(productos[2:])    # desde la posición 2 hasta el final
````

*Salida en el libro:*

```text
['camisa', 'jean']
['camisa', 'jean', 'chaqueta']
['chaqueta', 'medias']
```

#### Cambiar una lista

````python
ventas_del_dia = [45000, 120000]

ventas_del_dia.append(35000)      # agrega al final
print(ventas_del_dia)

ventas_del_dia[0] = 47000         # reemplaza el elemento de la posición 0
print(ventas_del_dia)

ventas_del_dia.remove(120000)     # quita el primer elemento con ese valor
print(ventas_del_dia)

ventas_del_dia.sort()             # ordena la lista de menor a mayor
print(ventas_del_dia)
````

*Salida en el libro:*

```text
[45000, 120000, 35000]
[47000, 120000, 35000]
[47000, 35000]
[35000, 47000]
```

### 6.3 Conjuntos: valores sin repetir

````python
ciudades_de_las_ventas = ["cali", "bogota", "cali", "medellin", "bogota",
                          "cali"]

# set() convierte la lista en un conjunto: los repetidos desaparecen.
ciudades_distintas = set(ciudades_de_las_ventas)

print(len(ciudades_distintas))
print("cali" in ciudades_distintas)
print("pasto" in ciudades_distintas)
````

*Salida en el libro:*

#### Buscar repetidos en una sola pasada

````python
codigos_pedidos = ["P01", "P02", "P03", "P02", "P04", "P05", "P03", "P06"]

codigos_vistos = set()
codigos_repetidos = set()

for codigo in codigos_pedidos:
    if codigo in codigos_vistos:
        codigos_repetidos.add(codigo)    # .add() agrega a un conjunto
    else:
        codigos_vistos.add(codigo)

print(sorted(codigos_repetidos))
````

*Salida en el libro:*

```text
['P02', 'P03']
```

#### Operaciones entre conjuntos

````python
clientes_enero = {"ana", "luis", "marta", "pedro"}
clientes_febrero = {"luis", "marta", "sofia"}

# & (intersección): los que están en los dos.
print(sorted(clientes_enero & clientes_febrero))

# | (unión): los que están en al menos uno.
print(sorted(clientes_enero | clientes_febrero))

# - (diferencia): los que están en el primero y no en el segundo.
print(sorted(clientes_enero - clientes_febrero))
````

*Salida en el libro:*

```text
['luis', 'marta']
['ana', 'luis', 'marta', 'pedro', 'sofia']
['ana', 'pedro']
```

### 6.4 Diccionarios: buscar un valor por su nombre

````python
# El precio de cada producto: el producto es la clave y el precio, el valor.
precios = {"camisa": 45000, "jean": 120000, "chaqueta": 230000}

# Se busca un valor por su clave, no por su posición.
print(precios["jean"])

# Agregar un par nuevo o cambiar uno existente se hace igual.
precios["medias"] = 15000
precios["camisa"] = 47000
print(precios)
````

*Salida en el libro:*

```text
120000
{'camisa': 47000, 'jean': 120000, 'chaqueta': 230000, 'medias': 15000}
```

````python
precios = {"camisa": 45000, "jean": 120000}

# .get() devuelve el segundo valor si la clave no existe.
print(precios.get("gorra", 0))
````

*Salida en el libro:*

```text
0
```

````python
precios = {"camisa": 45000, "jean": 120000, "chaqueta": 230000}

for producto, precio in precios.items():
    print(f"{producto}: {precio} pesos")
````

*Salida en el libro:*

```text
camisa: 45000 pesos
jean: 120000 pesos
chaqueta: 230000 pesos
```

#### El uso estrella: agrupar y sumar

````python
ventas = [("cali", 45000), ("bogota", 120000), ("cali", 35000),
          ("medellin", 89900), ("bogota", 15500)]

ventas_por_ciudad = {}
for ciudad, valor in ventas:
    # Si la ciudad aún no está, .get() devuelve 0; después sumamos el valor.
    ventas_por_ciudad[ciudad] = ventas_por_ciudad.get(ciudad, 0) + valor

print(ventas_por_ciudad)
````

*Salida en el libro:*

```text
{'cali': 80000, 'bogota': 135500, 'medellin': 89900}
```

### 6.7 Comprensiones de diccionarios y de conjuntos

````python
precios = {"camisa": 45000, "jean": 120000, "chaqueta": 230000}

# Comprensión de diccionario: {clave: valor for ...}
precios_con_iva = {producto: round(precio * 1.19)
                   for producto, precio in precios.items()}
print(precios_con_iva)

# Comprensión de conjunto: {valor for ...}
ciudades_crudas = ["Cali ", "cali", " CALI", "Bogota"]
ciudades_limpias = {ciudad.strip().lower() for ciudad in ciudades_crudas}
print(sorted(ciudades_limpias))
````

*Salida en el libro:*

```text
{'camisa': 53550, 'jean': 142800, 'chaqueta': 273700}
['bogota', 'cali']
```

### [Ex-L1-C6-02] — Repetidos y clientes que no volvieron

````python
codigos_pedidos = ["P10", "P11", "P12", "P11", "P13", "P14", "P10",
                   "P15", "P11"]
clientes_marzo = {"ana", "luis", "marta", "pedro", "sofia"}
clientes_abril = {"luis", "sofia", "carlos", "diana"}
````

````python
codigos_pedidos = ["P10", "P11", "P12", "P11", "P13", "P14", "P10",
                   "P15", "P11"]
clientes_marzo = {"ana", "luis", "marta", "pedro", "sofia"}
clientes_abril = {"luis", "sofia", "carlos", "diana"}

# --- 1. Repetidos con un solo recorrido ---
codigos_vistos = set()
codigos_repetidos = set()
for codigo in codigos_pedidos:
    if codigo in codigos_vistos:
        codigos_repetidos.add(codigo)
    else:
        codigos_vistos.add(codigo)
print(f"Códigos repetidos: {sorted(codigos_repetidos)}")

# --- 2. Pedidos distintos, en una línea ---
print(f"Pedidos distintos: {len(set(codigos_pedidos))}")

# --- 3. Comparación de clientes entre meses ---
compraron_ambos = sorted(clientes_marzo & clientes_abril)
print(f"Compraron en los dos meses: {compraron_ambos}")
print(f"No volvieron en abril: {sorted(clientes_marzo - clientes_abril)}")
print(f"Nuevos en abril: {sorted(clientes_abril - clientes_marzo)}")

# --- 4. Clientes distintos en total ---
print(f"Clientes distintos: {len(clientes_marzo | clientes_abril)}")
````

*Salida en el libro:*

```text
Códigos repetidos: ['P10', 'P11']
Pedidos distintos: 6
Compraron en los dos meses: ['luis', 'sofia']
No volvieron en abril: ['ana', 'marta', 'pedro']
Nuevos en abril: ['carlos', 'diana']
Clientes distintos: 7
```

### [Ex-L1-C6-03] — Ventas por categoría

````python
ventas = [("ropa", 45000), ("hogar", 120000), ("ropa", 35000),
          ("electronica", 390000), ("hogar", 15500), ("ropa", 67000),
          ("ropa", 89000), ("hogar", 52000), ("ropa", 120000),
          ("ropa", 99000)]
````

````python
ventas = [("ropa", 45000), ("hogar", 120000), ("ropa", 35000),
          ("electronica", 390000), ("hogar", 15500), ("ropa", 67000),
          ("ropa", 89000), ("hogar", 52000), ("ropa", 120000),
          ("ropa", 99000)]

# --- 1 y 2. Total y número de ventas por categoría, en un solo recorrido ---
total_por_categoria = {}
numero_por_categoria = {}
for categoria, valor in ventas:
    total_por_categoria[categoria] = (
        total_por_categoria.get(categoria, 0) + valor)
    numero_por_categoria[categoria] = (
        numero_por_categoria.get(categoria, 0) + 1)

print("Total por categoría:")
print(total_por_categoria)
print("Número de ventas por categoría:")
print(numero_por_categoria)

# --- 3. Promedio por categoría ---
for categoria, total in total_por_categoria.items():
    promedio = total / numero_por_categoria[categoria]
    print(f"Promedio de {categoria}: {promedio:.0f} pesos")
````

### [Ex-L1-C6-04] — ¿Qué tan parecidos son dos clientes?

````python
compras_ana = {"camisa", "jean", "medias", "gorra"}
compras_luis = {"camisa", "jean", "chaqueta"}
compras_marta = {"sarten", "olla", "medias"}
````

````python
compras_ana = {"camisa", "jean", "medias", "gorra"}
compras_luis = {"camisa", "jean", "chaqueta"}
compras_marta = {"sarten", "olla", "medias"}

# --- 1. Índice de Jaccard para cada par ---
jaccard_ana_luis = (len(compras_ana & compras_luis)
                    / len(compras_ana | compras_luis))
jaccard_ana_marta = (len(compras_ana & compras_marta)
                     / len(compras_ana | compras_marta))
jaccard_luis_marta = (len(compras_luis & compras_marta)
                      / len(compras_luis | compras_marta))

print(f"Ana y Luis: {jaccard_ana_luis:.2f}")
print(f"Ana y Marta: {jaccard_ana_marta:.2f}")
print(f"Luis y Marta: {jaccard_luis_marta:.2f}")

# --- 2. Recomendación para Luis: lo que compró Ana y Luis todavía no ---
recomendaciones_luis = compras_ana - compras_luis
print(f"Recomendar a Luis: {sorted(recomendaciones_luis)}")
````

*Salida en el libro:*

```text
Ana y Luis: 0.40
Ana y Marta: 0.17
Luis y Marta: 0.00
Recomendar a Luis: ['gorra', 'medias']
```

````python
compras_cliente_nuevo_1 = set()
compras_cliente_nuevo_2 = set()

union_compras = compras_cliente_nuevo_1 | compras_cliente_nuevo_2
if len(union_compras) == 0:
    jaccard = 0.0    # sin compras no hay base para decir que se parecen
else:
    jaccard = (len(compras_cliente_nuevo_1 & compras_cliente_nuevo_2)
               / len(union_compras))
print(f"Índice de Jaccard: {jaccard:.2f}")
````

*Salida en el libro:*

```text
Índice de Jaccard: 0.00
```

### [Ex-L1-C6-05] — De filas a columnas

````python
pedidos = [
    {"codigo": "P01", "producto": "camisa", "ciudad": "cali",
     "valor": 45000},
    {"codigo": "P02", "producto": "jean", "ciudad": "bogota",
     "valor": 120000},
    {"codigo": "P03", "producto": "chaqueta", "ciudad": "cali",
     "valor": 230000},
    {"codigo": "P04", "producto": "medias", "ciudad": "medellin",
     "valor": 15000},
    {"codigo": "P05", "producto": "camisa", "ciudad": "bogota",
     "valor": 45000},
]
````

````python
pedidos = [
    {"codigo": "P01", "producto": "camisa", "ciudad": "cali",
     "valor": 45000},
    {"codigo": "P02", "producto": "jean", "ciudad": "bogota",
     "valor": 120000},
    {"codigo": "P03", "producto": "chaqueta", "ciudad": "cali",
     "valor": 230000},
    {"codigo": "P04", "producto": "medias", "ciudad": "medellin",
     "valor": 15000},
    {"codigo": "P05", "producto": "camisa", "ciudad": "bogota",
     "valor": 45000},
]

# --- 1. Pedidos de Cali (cómodo con filas: cada pedido está completo) ---
for pedido in pedidos:
    if pedido["ciudad"] == "cali":
        print(f"{pedido['codigo']}: {pedido['valor']} pesos")

# --- 2. De filas a columnas ---
pedidos_por_columnas = {"codigo": [], "producto": [], "ciudad": [],
                        "valor": []}
for pedido in pedidos:
    for columna in pedidos_por_columnas:
        pedidos_por_columnas[columna].append(pedido[columna])

print(pedidos_por_columnas["valor"])

# --- 3. Total y promedio (cómodo con columnas: la columna ya está junta) ---
valores = pedidos_por_columnas["valor"]
print(f"Valor total: {sum(valores)} pesos")
print(f"Valor promedio: {sum(valores) / len(valores):.0f} pesos")
````

*Salida en el libro:*

```text
P01: 45000 pesos
P03: 230000 pesos
[45000, 120000, 230000, 15000, 45000]
Valor total: 455000 pesos
Valor promedio: 91000 pesos
```
