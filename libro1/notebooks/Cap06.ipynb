### 6.1 Listas: la colección ordenada y mutable

```python
# Una lista de montos de transacciones. Ordenada y modificable.
montos = [150.0, 2300.0, 75.5, 9800.0]
 
print(montos[0])    # 150.0  -> primer elemento
print(montos[-1])   # 9800.0 -> último elemento (índice negativo)
 
# Slicing: extraer una sub-lista con [inicio:fin] (fin excluido).
print(montos[1:3])  # [2300.0, 75.5]
```

### 6.2 Tuplas: la colección ordenada e inmutable

```python
# Una tupla representa datos que NO deben cambiar (coordenadas de una
# sucursal).
coordenadas_sucursal = (10.3910, -75.4794)  # (latitud, longitud) de Cartagena
 
# Desempaquetado (unpacking): una variable por elemento, de una sola vez.
latitud, longitud = coordenadas_sucursal
print(f"Latitud: {latitud}, Longitud: {longitud}")
```

### 6.3 Sets: la colección de elementos únicos

```python
# Un set elimina automáticamente los duplicados.
metodos_pago = {"tarjeta", "transferencia", "tarjeta", "cripto"}
print(metodos_pago)  # {'tarjeta', 'transferencia', 'cripto'}
 
vocabulario_espanol = {"casa", "sol", "agua", "luz", "mar"}
vocabulario_italiano = {"casa", "sole", "acqua", "luce", "mare"}
 
# Operaciones de conjuntos.
interseccion = vocabulario_espanol & vocabulario_italiano  # {'casa'}
union = vocabulario_espanol | vocabulario_italiano
diferencia = vocabulario_espanol - vocabulario_italiano
```

### 6.4 Diccionarios: pares clave-valor

```python
# Un diccionario modela un registro con campos nombrados.
cliente = {
    "nombre": "Ana García",
    "edad": 34,
    "ciudad": "Cartagena",
    "es_premium": True,
}
 
print(cliente["nombre"])  # acceso por clave
 
# .get() no falla si la clave no existe; devuelve un valor por defecto.
telefono = cliente.get("telefono", "No registrado")
 
# .items() entrega pares (clave, valor) para iterar.
for clave, valor in cliente.items():
    print(f"{clave}: {valor}")
```

### 6.6 Conversión entre estructuras

```python
# Eliminar duplicados convirtiendo a set y de vuelta a lista.
montos_con_repetidos = [100.0, 250.0, 100.0, 80.0, 250.0]
montos_unicos = list(set(montos_con_repetidos))
print(montos_unicos)  # orden no garantizado
```

### 6.7 Comprehensions de diccionarios y de conjuntos

```python
# Dict comprehension: mapear cada producto a su precio con IVA.
precios_base = {"cafe": 10.0, "te": 8.0, "agua": 5.0}
TASA_IVA = 0.19
 
precios_con_iva = {producto: precio * (1 + TASA_IVA)
                   for producto, precio in precios_base.items()}
print(precios_con_iva)
```

### 6.8 Puente hacia Pandas: representar un dataset tabular

```python
# Enfoque orientado a FILAS: una lista de diccionarios.
clientes_por_filas = [
    {"nombre": "Ana",   "edad": 34, "ciudad": "Cartagena"},
    {"nombre": "Luis",  "edad": 28, "ciudad": "Bogotá"},
    {"nombre": "Marta", "edad": 45, "ciudad": "Medellín"},
]
 
# Edad promedio recorriendo los registros:
suma_edades = 0
for cliente in clientes_por_filas:
    suma_edades += cliente["edad"]
edad_promedio = suma_edades / len(clientes_por_filas)
```

```python
# Enfoque orientado a COLUMNAS: un diccionario de listas.
clientes_por_columnas = {
    "nombre": ["Ana", "Luis", "Marta"],
    "edad":   [34, 28, 45],
    "ciudad": ["Cartagena", "Bogotá", "Medellín"],
}
 
# La columna ya está disponible para operar de un golpe.
edades = clientes_por_columnas["edad"]
edad_promedio = sum(edades) / len(edades)
```

### [Ex-L1-C6-01] — Gestión de inventario

```python
"""Gestión de inventario: análisis sobre una lista de diccionarios
(registros)."""
 
# --- Constante de política ---
UMBRAL_PRECIO_PREMIUM = 100.0
 
# --- El inventario: una lista de diccionarios (orientado a registros) ---
# Cada diccionario es una FILA; las claves son las COLUMNAS.
inventario = [
    {"nombre": "Teclado mecánico",  "precio": 120.0, "stock": 15},
    {"nombre": "Mouse inalámbrico", "precio": 45.5,  "stock": 40},
    {"nombre": "Monitor 27\"",      "precio": 310.0, "stock": 8},
    {"nombre": "Webcam HD",         "precio": 60.0,  "stock": 25},
    {"nombre": "Auriculares",       "precio": 85.0,  "stock": 30},
]
 
# --- 1. Productos premium (precio > umbral) con list comprehension ---
productos_premium = [
    producto["nombre"]
    for producto in inventario
    if producto["precio"] > UMBRAL_PRECIO_PREMIUM
]
print(f"Productos premium: {productos_premium}")
 
# --- 2. Valor total del inventario (suma de precio * stock) ---
valor_total = 0.0
for producto in inventario:
    # Acceso seguro con .get(): si faltara 'stock', devolvería 0 en vez de
    # fallar.
    unidades = producto.get("stock", 0)
    valor_total += producto["precio"] * unidades
print(f"Valor total del inventario: ${valor_total:,.2f}")
 
# --- 3. Producto con mayor stock (campeón provisional) ---
producto_mayor_stock = inventario[0]  # campeón inicial: el primero
for producto in inventario:
    if producto["stock"] > producto_mayor_stock["stock"]:
        producto_mayor_stock = producto
print(
    f"Mayor stock: {producto_mayor_stock['nombre']} "
    f"({producto_mayor_stock['stock']} unidades)"
)
```

### [Ex-L1-C6-02] — Recuento de votos

```python
"""Recuento de votos: candidatos únicos con set, conteo con diccionario."""
 
# Lista de votos emitidos (con repeticiones).
votos = [
    "Ana", "Luis", "Ana", "Marta", "Luis", "Ana",
    "Marta", "Ana", "Luis", "Marta", "Marta", "Ana",
]
 
# --- 1. Candidatos únicos usando un set ---
# El set elimina automáticamente los duplicados: unicidad sin esfuerzo.
candidatos_unicos = set(votos)
print(f"Candidatos: {sorted(candidatos_unicos)}")
 
# --- 2. Conteo de votos con un diccionario ---
conteo_votos = {}
for voto in votos:
    # .get(voto, 0) devuelve el conteo actual, o 0 si es su primer voto.
    # Búsqueda y actualización en O(1) gracias al hash.
    conteo_votos[voto] = conteo_votos.get(voto, 0) + 1
print(f"Conteo: {conteo_votos}")
 
# --- 3. Determinar el ganador (campeón provisional sobre el diccionario) ---
candidato_ganador = ""
votos_ganador = 0
for candidato, total in conteo_votos.items():
    if total > votos_ganador:
        candidato_ganador = candidato
        votos_ganador = total
 
# --- 4. Reto: detectar empate en el primer lugar ---
candidatos_en_maximo = [
    candidato
    for candidato, total in conteo_votos.items()
    if total == votos_ganador
]
 
if len(candidatos_en_maximo) > 1:
    print(f"¡Empate! entre {candidatos_en_maximo} con {votos_ganador} votos.")
else:
    print(f"Ganador: {candidato_ganador} con {votos_ganador} votos.")
```

### [Ex-L1-C6-03] — Similitud de idiomas I: el Índice de Jaccard con conjuntos

```python
"""Similitud de idiomas I: Índice de Jaccard entre vocabularios usando sets."""
 
# --- Vocabularios como sets, con cognados realistas entre lenguas romances ---
vocabulario_espanol = {
    "casa", "sol", "gato", "flor", "mano", "libro", "agua", "verde",
}
vocabulario_italiano = {
    "casa", "sole", "gatto", "fiore", "mano", "libro", "acqua", "verde",
}
vocabulario_portugues = {
    "casa", "sol", "gato", "flor", "mao", "livro", "agua", "verde",
}
 
# --- Cálculo del Índice de Jaccard usando SOLO operaciones de conjuntos ---
# & -> intersección (elementos comunes); | -> unión (todos los elementos).
 
# Par 1: español vs italiano
interseccion_esp_ita = vocabulario_espanol & vocabulario_italiano
union_esp_ita = vocabulario_espanol | vocabulario_italiano
jaccard_esp_ita = len(interseccion_esp_ita) / len(union_esp_ita)
 
# Par 2: español vs portugués
interseccion_esp_por = vocabulario_espanol & vocabulario_portugues
union_esp_por = vocabulario_espanol | vocabulario_portugues
jaccard_esp_por = len(interseccion_esp_por) / len(union_esp_por)
 
# Par 3: italiano vs portugués
interseccion_ita_por = vocabulario_italiano & vocabulario_portugues
union_ita_por = vocabulario_italiano | vocabulario_portugues
jaccard_ita_por = len(interseccion_ita_por) / len(union_ita_por)
 
# --- Informe con valores redondeados a 3 decimales ---
print(f"Jaccard español-italiano:   {jaccard_esp_ita:.3f}")
print(f"Jaccard español-portugués:  {jaccard_esp_por:.3f}")
print(f"Jaccard italiano-portugués: {jaccard_ita_por:.3f}")
 
# --- Determinación del par más similar (campeón provisional) ---
mayor_similitud = jaccard_esp_ita
par_mas_similar = "español-italiano"
 
if jaccard_esp_por > mayor_similitud:
    mayor_similitud = jaccard_esp_por
    par_mas_similar = "español-portugués"
if jaccard_ita_por > mayor_similitud:
    mayor_similitud = jaccard_ita_por
    par_mas_similar = "italiano-portugués"
 
print(f"\nPar más similar: {par_mas_similar} ({mayor_similitud:.3f})")
```
