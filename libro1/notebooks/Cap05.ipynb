### 5.1 Repetir con cada elemento: el bucle for

````python
ventas_del_dia = [45000, 120000, 35000, 89900]

# En cada vuelta, la variable venta toma el siguiente valor de la lista.
for venta in ventas_del_dia:
    print(f"Revisando una venta de {venta} pesos")
````

*Salida en el libro:*

```text
Revisando una venta de 45000 pesos
Revisando una venta de 120000 pesos
Revisando una venta de 35000 pesos
Revisando una venta de 89900 pesos
```

#### Sumar y contar paso a paso

````python
ventas_del_dia = [45000, 120000, 35000, 89900, 15500]

total_vendido = 0          # el acumulador empieza en cero
ventas_grandes = 0         # el contador también

for venta in ventas_del_dia:
    total_vendido += venta              # sumamos cada venta
    if venta >= 50000:
        ventas_grandes += 1             # contamos solo las grandes

print(f"Total vendido: {total_vendido} pesos")
print(f"Ventas de 50000 pesos o más: {ventas_grandes}")
````

*Salida en el libro:*

```text
Total vendido: 305400 pesos
Ventas de 50000 pesos o más: 2
```

#### Repetir un número de veces: range()

#### Numerar los elementos: enumerate()

````python
productos_mas_vendidos = ["camisa", "jean", "chaqueta"]

# enumerate() entrega, en cada vuelta, la posición y el elemento.
# start=1 hace que la numeración empiece en 1 y no en 0.
for posicion, producto in enumerate(productos_mas_vendidos, start=1):
    print(f"{posicion}. {producto}")
````

*Salida en el libro:*

```text
1. camisa
2. jean
3. chaqueta
```

#### Recorrer dos listas a la vez: zip()

````python
ciudad = "bogotá"

vocales_con_tilde = "áéíóú"
vocales_sin_tilde = "aeiou"

# En cada vuelta, zip() entrega una vocal con tilde y su versión sin tilde.
for con_tilde, sin_tilde in zip(vocales_con_tilde, vocales_sin_tilde):
    ciudad = ciudad.replace(con_tilde, sin_tilde)

print(ciudad)
````

*Salida en el libro:*

```text
bogota
```

### 5.2 Repetir mientras se cumpla algo: el bucle while

````python
ventas_mensuales = 10000000
META_VENTAS = 15000000
CRECIMIENTO_MENSUAL = 0.05

meses_transcurridos = 0

# Mientras no se alcance la meta, avanzamos un mes.
while ventas_mensuales < META_VENTAS:
    ventas_mensuales *= 1 + CRECIMIENTO_MENSUAL   # las ventas crecen 5 %
    meses_transcurridos += 1                       # contamos el mes

print(f"La meta se alcanza en {meses_transcurridos} meses.")
print(f"Ventas de ese mes: {ventas_mensuales:.0f} pesos")
````

*Salida en el libro:*

```text
La meta se alcanza en 9 meses.
Ventas de ese mes: 15513282 pesos
```

### 5.3 Detener o saltar una vuelta: break y continue

````python
ventas_registradas = [45000, -5000, 120000, 0, 35000]

total_valido = 0
for venta in ventas_registradas:
    if venta <= 0:
        print(f"Se omite un valor imposible: {venta}")
        continue                  # salta a la siguiente venta
    total_valido += venta

print(f"Total de las ventas válidas: {total_valido} pesos")
````

*Salida en el libro:*

```text
Se omite un valor imposible: -5000
Se omite un valor imposible: 0
Total de las ventas válidas: 200000 pesos
```

````python
ventas_registradas = [45000, 120000, 2300000, 35000, 1800000]
VENTA_MUY_GRANDE = 1000000

for venta in ventas_registradas:
    if venta >= VENTA_MUY_GRANDE:
        print(f"Revisar: se encontró una venta de {venta} pesos.")
        break                     # detiene la búsqueda
else:
    # Este bloque se ejecuta solo si el bucle terminó sin hacer break.
    print("Ninguna venta supera el límite.")
````

*Salida en el libro:*

```text
Revisar: se encontró una venta de 2300000 pesos.
```

### 5.5 Una forma corta de crear listas: las comprensiones

````python
precios_sin_iva = [45000, 120000, 35000]

# Forma larga: lista vacía, bucle y append().
precios_con_iva = []
for precio in precios_sin_iva:
    precios_con_iva.append(round(precio * 1.19))

print(precios_con_iva)
````

*Salida en el libro:*

```text

```

````python
precios_sin_iva = [45000, 120000, 35000]

# Forma corta: todo en una línea, entre corchetes.
precios_con_iva = [round(precio * 1.19) for precio in precios_sin_iva]

print(precios_con_iva)
````

*Salida en el libro:*


### [Ex-L1-C5-01] — Resumen de ventas paso a paso

````python
ventas_semana = [45000, 120000, 35000, 89900, 15500, 230000, 67000]
````

````python
ventas_semana = [45000, 120000, 35000, 89900, 15500, 230000, 67000]
UMBRAL_VENTA_ALTA = 100000

# --- Valores iniciales ---
total_vendido = 0                  # acumulador
numero_ventas = 0                  # contador de todas las ventas
ventas_sobre_umbral = 0            # contador de ventas altas
venta_mas_alta = ventas_semana[0]  # campeón provisional: la primera venta

# --- Un solo recorrido calcula todo ---
for venta in ventas_semana:
    total_vendido += venta
    numero_ventas += 1
    if venta > venta_mas_alta:
        venta_mas_alta = venta     # aparece un nuevo campeón
    if venta > UMBRAL_VENTA_ALTA:
        ventas_sobre_umbral += 1

print(f"Total vendido: {total_vendido} pesos")
print(f"Número de ventas: {numero_ventas}")
print(f"Venta más alta: {venta_mas_alta} pesos")
print(f"Ventas de más de {UMBRAL_VENTA_ALTA} pesos: {ventas_sobre_umbral}")

# --- Comprobación con las funciones incorporadas ---
print(total_vendido == sum(ventas_semana))
print(numero_ventas == len(ventas_semana))
print(venta_mas_alta == max(ventas_semana))
````

*Salida en el libro:*

```text
Total vendido: 602400 pesos
Número de ventas: 7
Venta más alta: 230000 pesos
Ventas de más de 100000 pesos: 2

True
True
```

### [Ex-L1-C5-02] — Limpiar una columna de ciudades

````python
ciudades_crudas = ["  Bogotá", "MEDELLÍN ", "cali", "Bogota", " medellín",
                   "CALI"]
````

````python
ciudades_crudas = ["  Bogotá", "MEDELLÍN ", "cali", "Bogota", " medellín",
                   "CALI"]

VOCALES_CON_TILDE = "áéíóú"
VOCALES_SIN_TILDE = "aeiou"

# --- 1. Limpieza de cada ciudad ---
ciudades_limpias = []
for ciudad in ciudades_crudas:
    ciudad = ciudad.strip().lower()
    for con_tilde, sin_tilde in zip(VOCALES_CON_TILDE, VOCALES_SIN_TILDE):
        ciudad = ciudad.replace(con_tilde, sin_tilde)
    ciudades_limpias.append(ciudad)

# --- 2. Resultado ---
print(ciudades_limpias)

# --- 3. Ventas en Bogotá antes y después de limpiar ---
ventas_bogota_antes = 0
for ciudad in ciudades_crudas:
    if ciudad == "Bogotá":
        ventas_bogota_antes += 1

ventas_bogota_despues = 0
for ciudad in ciudades_limpias:
    if ciudad == "bogota":
        ventas_bogota_despues += 1

print(f"Ventas en Bogotá antes de limpiar: {ventas_bogota_antes}")
print(f"Ventas en Bogotá después de limpiar: {ventas_bogota_despues}")
````

*Salida en el libro:*

```text
['bogota', 'medellin', 'cali', 'bogota', 'medellin', 'cali']
Ventas en Bogotá antes de limpiar: 0
Ventas en Bogotá después de limpiar: 2
```

### [Ex-L1-C5-03] — Revisión de pedidos con break y continue

````python
valores_pedidos = [45000, 0, 120000, -30000, 89900, 15500, 3500000, 67000]
````

````python
valores_pedidos = [45000, 0, 120000, -30000, 89900, 15500, 3500000, 67000]
PEDIDO_MUY_GRANDE = 3000000

# --- 1. Total de pedidos válidos, saltando los imposibles ---
total_valido = 0
valores_omitidos = 0
for valor in valores_pedidos:
    if valor <= 0:
        valores_omitidos += 1
        continue
    total_valido += valor

print(f"Total de pedidos válidos: {total_valido} pesos")
print(f"Valores imposibles omitidos: {valores_omitidos}")

# --- 2. Primer pedido muy grande ---
for posicion, valor in enumerate(valores_pedidos, start=1):
    if valor >= PEDIDO_MUY_GRANDE:
        print(f"Pedido muy grande en la posición {posicion}: {valor} pesos")
        break
else:
    print(f"Ningún pedido llega a {PEDIDO_MUY_GRANDE} pesos.")
````


### [Ex-L1-C5-04] — Buscar pedidos repetidos

````python
codigos_pedidos = ["P01", "P02", "P03", "P02", "P04", "P05", "P03", "P06"]
````

````python
codigos_pedidos = ["P01", "P02", "P03", "P02", "P04", "P05", "P03", "P06"]

numero_comparaciones = 0
for i in range(len(codigos_pedidos)):
    # Solo se compara con los códigos que están después de la posición i,
    # para no comparar dos veces el mismo par ni un código consigo mismo.
    for j in range(i + 1, len(codigos_pedidos)):
        numero_comparaciones += 1
        if codigos_pedidos[i] == codigos_pedidos[j]:
            print(f"Repetido: {codigos_pedidos[i]} "
                  f"(posiciones {i} y {j})")

print(f"Comparaciones realizadas: {numero_comparaciones}")
````

*Salida en el libro:*

```text
Repetido: P02 (posiciones 1 y 3)
Repetido: P03 (posiciones 2 y 6)
Comparaciones realizadas: 28
```

### [Ex-L1-C5-05] — Agrupar ventas por rangos

````python
ventas_mes = [45000, 120000, 35000, 89900, 15500, 230000, 67000,
              310000, 52000, 18000, 99000, 150000, 27000, 75000]
````

*Texto del libro:*

```text
Pequeñas  ****
```

````python
ventas_mes = [45000, 120000, 35000, 89900, 15500, 230000, 67000,
              310000, 52000, 18000, 99000, 150000, 27000, 75000]

LIMITE_MEDIANA = 50000
LIMITE_GRANDE = 150000

# --- 1. Tres grupos con comprensiones de listas ---
ventas_pequenas = [venta for venta in ventas_mes if venta < LIMITE_MEDIANA]
ventas_medianas = [venta for venta in ventas_mes
                   if LIMITE_MEDIANA <= venta < LIMITE_GRANDE]
ventas_grandes = [venta for venta in ventas_mes if venta >= LIMITE_GRANDE]

# --- 2. Cantidad y suma de cada grupo ---
print(f"Pequeñas: {len(ventas_pequenas)} ventas, "
      f"{sum(ventas_pequenas)} pesos")
print(f"Medianas: {len(ventas_medianas)} ventas, "
      f"{sum(ventas_medianas)} pesos")
print(f"Grandes:  {len(ventas_grandes)} ventas, "
      f"{sum(ventas_grandes)} pesos")

# --- 3. Gráfico de barras en texto ---
print()
print(f"Pequeñas  {'*' * len(ventas_pequenas)}")
print(f"Medianas  {'*' * len(ventas_medianas)}")
print(f"Grandes   {'*' * len(ventas_grandes)}")

# --- 4. Comprobación: ninguna venta se perdió ni se contó dos veces ---
total_agrupado = (len(ventas_pequenas) + len(ventas_medianas)
                  + len(ventas_grandes))
print()
print(f"Ventas agrupadas: {total_agrupado} de {len(ventas_mes)}")
````

*Salida en el libro:*

```text
Pequeñas: 5 ventas, 140500 pesos
Medianas: 6 ventas, 502900 pesos
Grandes:  3 ventas, 690000 pesos

Pequeñas  *****
Medianas  ******
Grandes   ***

Ventas agrupadas: 14 de 14
```
