### 3.6 Built-in functions esenciales

````python
# len() -> longitud (número de elementos o caracteres).
nombre_cliente = "Ana García"
print(len(nombre_cliente))  # 10 (incluye el espacio)
 
# abs() -> valor absoluto.
variacion_saldo = -350.75
print(abs(variacion_saldo))  # 350.75
 
# round() -> redondea a los decimales indicados.
tasa_conversion = 0.073456
print(round(tasa_conversion, 3))  # 0.073
 
# range() -> genera una secuencia de enteros (clave para los bucles).
print(list(range(5)))  # [0, 1, 2, 3, 4]
 
# sorted() -> versión ordenada de una secuencia.
montos = [150.5, 20.0, 99.9, 5.25]
print(sorted(montos))  # [5.25, 20.0, 99.9, 150.5]
````

### [Ex-L1-C3-01] — Script de presentación del cliente

````python
"""Script de presentación: recoge datos del usuario y calcula su año de
nacimiento."""
 
from datetime import date
 
# Año en curso, tomado del reloj del sistema: el cálculo no caduca.
ANIO_ACTUAL = date.today().year
 
# --- Entrada de datos ---
# input() devuelve SIEMPRE un str, por lo que nombre y ciudad ya son texto.
nombre_cliente = input("Introduzca su nombre: ")
ciudad_residencia = input("Introduzca su ciudad: ")
edad_texto = input("Introduzca su edad: ")
 
# --- Conversión de tipo (paso CRÍTICO) ---
# La edad llega como texto; la convertimos a int para poder restarla.
edad_cliente = int(edad_texto)
 
# --- Cálculo ---
# Año de nacimiento aproximado: restamos la edad al año actual.
anio_nacimiento_aprox = ANIO_ACTUAL - edad_cliente
 
# --- Salida formateada con f-string ---
print(
    f"Hola {nombre_cliente}, de {ciudad_residencia}. "
    f"Nació aproximadamente en {anio_nacimiento_aprox}."
)
````

### Versión corregida y conforme a PEP 8

````python
"""Versión corregida: suma de dos valores numéricos y redondeo."""
 
# Error 1 corregido: nombre en snake_case, sin espacios.
# Error 2 corregido: convertimos los textos a números ANTES de operar.
edad_cliente = int("28")          # de str a int
monto = float("1500.75")          # de str a float
 
# Ahora la suma es aritmética y produce un número real.
total = edad_cliente + monto      # 28 + 1500.75 = 1528.75
 
# Errores 4 y 5 corregidos: round() sobre un valor numérico.
resultado = round(total, 2)
 
# Error 3 corregido: f-string real, con 'f' inicial y variable entre llaves.
print(f"El total es: {resultado}")
````
