### [Ex-L1-C3-01] — Un pedido que llegó como texto

````python
codigo_pedido = "PED-0007"
producto = "Chaqueta impermeable"
precio_unitario_texto = "120000"
cantidad_texto = "2"
descuento_texto = "0.10"
````

````python
# --- Datos del pedido, tal como llegaron de la página web (todo es texto) ---
codigo_pedido = "PED-0007"
producto = "Chaqueta impermeable"
precio_unitario_texto = "120000"
cantidad_texto = "2"
descuento_texto = "0.10"

# --- 1. Conversión de tipos ---
# El precio y la cantidad no tienen decimales: los convertimos en enteros.
precio_unitario = int(precio_unitario_texto)
cantidad = int(cantidad_texto)
# El descuento sí tiene decimales: lo convertimos en número con decimales.
descuento = float(descuento_texto)

# --- 2. Cálculo del total ---
# Con un descuento del 10 %, el cliente paga el 90 %:
# se multiplica por (1 - 0.10).
total_pagado = precio_unitario * cantidad * (1 - descuento)
print(f"Pedido {codigo_pedido}: {cantidad} x {producto}")
print(f"Total pagado: {total_pagado:.0f} pesos")

# --- 3. Tipo de cada variable convertida ---
print(type(precio_unitario))
print(type(cantidad))
print(type(descuento))
print(type(total_pagado))
````

*Salida en el libro:*

```text
Pedido PED-0007: 2 x Chaqueta impermeable
Total pagado: 216000 pesos
<class 'int'>
<class 'int'>
<class 'float'>
<class 'float'>
```

### [Ex-L1-C3-03] — Cazador de errores

*Código con errores a propósito (no se ejecuta):*

```python
precio_unitario = "45000"
cantidad = "3"
costo_envio = "8000"

total_productos = precio_unitario * cantidad          # línea A
cantidad_con_regalo = cantidad + 1                    # línea B
descuento = int("0.15")                               # línea C
total_con_envio = str(135000) + costo_envio           # línea D
print(f"Total a pagar: {total_con_envio}")
```

````python
# --- Datos del pedido (llegaron como texto) ---
precio_unitario_texto = "45000"
cantidad_texto = "3"
costo_envio_texto = "8000"

# --- Conversión: primero convertimos, después calculamos ---
precio_unitario = int(precio_unitario_texto)
cantidad = int(cantidad_texto)
costo_envio = int(costo_envio_texto)

total_productos = precio_unitario * cantidad          # línea A corregida
cantidad_con_regalo = cantidad + 1                    # línea B corregida
descuento = float("0.15")                             # línea C corregida
total_con_envio = total_productos + costo_envio       # línea D corregida

print(f"Total a pagar: {total_con_envio}")
print(f"Unidades con regalo: {cantidad_con_regalo}")
print(f"Descuento guardado: {descuento}")
````

*Salida en el libro:*

```text
Total a pagar: 143000
Unidades con regalo: 4
Descuento guardado: 0.15
```

### [Ex-L1-C3-04] — Informe de ventas del día

````python
ventas_del_dia = [45000, 120000, 35000, 89900, 15500]
````

*Texto del libro:*

```text
INFORME DE VENTAS DEL DÍA
Número de ventas: ...
Total vendido: ... pesos
Venta promedio: ... pesos
Venta más alta: ... pesos
Venta más baja: ... pesos
La venta más alta representa el ...% del total
```

````python
# --- Ventas del día, en pesos ---
ventas_del_dia = [45000, 120000, 35000, 89900, 15500]

# --- Cálculos: cada valor sale de una función, nunca se escribe a mano ---
numero_ventas = len(ventas_del_dia)
total_vendido = sum(ventas_del_dia)
venta_promedio = total_vendido / numero_ventas
venta_mas_alta = max(ventas_del_dia)
venta_mas_baja = min(ventas_del_dia)
porcentaje_venta_mas_alta = venta_mas_alta / total_vendido * 100

# --- Informe ---
print("INFORME DE VENTAS DEL DÍA")
print(f"Número de ventas: {numero_ventas}")
print(f"Total vendido: {total_vendido} pesos")
print(f"Venta promedio: {venta_promedio:.0f} pesos")
print(f"Venta más alta: {venta_mas_alta} pesos")
print(f"Venta más baja: {venta_mas_baja} pesos")
print(f"La venta más alta representa el "
      f"{porcentaje_venta_mas_alta:.1f}% del total")
````

*Salida en el libro:*

```text
INFORME DE VENTAS DEL DÍA
Número de ventas: 5
Total vendido: 305400 pesos
Venta promedio: 61080 pesos
Venta más alta: 120000 pesos
Venta más baja: 15500 pesos
La venta más alta representa el 39.3% del total
```
